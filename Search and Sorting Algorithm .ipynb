### Linear Search


```python
def liner_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1
arr = [10, 50, 20, 30]
target = 20
result = liner_search(arr, target)
if result != 1 :
    print("element  is found", result)
else :
    print("element is not fount")
```

### Binary Search


```python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1
    while left <= right :
        mid = left + right // 2
        if arr[mid] == target :
            return mid
        elif arr[mid] < target :
            left = mid + 1
        else :
            right = mid-1
    return -1 
arr = [10, 12, 15, 20, 30, 40, 50]
target = 20
result = binary_search(arr, target)

if result != 1 :
    print("element  is found", result)
else :
    print("element is not fount")
```
